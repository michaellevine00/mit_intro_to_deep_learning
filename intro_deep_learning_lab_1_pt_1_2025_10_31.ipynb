## Math notation - Latex
For a linear layer:  
$$ y = \sigma(Wx + b) $$  
where  
- $ W \in \mathbb{R}^{d_{\text{out}} \times d_{\text{in}}} $ is the weight matrix  
- $ x \in \mathbb{R}^{d_{\text{in}}} $ is the input vector  
- $ b \in \mathbb{R}^{d_{\text{out}}} $ is the bias vector  
- $ \sigma(\cdot) $ is the activation function (e.g., ReLU, tanh, etc.)

We can also write it more compactly as  
$$ y = \sigma(Wx + b) $$

# 1.3 Neural Networks in PyTorch
We can also define neural networks in PyTorch. PyTorch uses torch.nn.Module, which serves as a base class for all neural network modules in PyTorch and thus provides a framework for building and training neural networks.

Let's consider the example of a simple perceptron defined by just one dense (aka fully-connected or linear) layer:$$y = \sigma(Wx + b)$$ where $$W$$ represents a matrix of weights, $$b$$ is a bias, $$x$$ is the input, $$\sigma$$ is the activation function, and $$y$$ is the output.

![neural_network_sigmoid_activation](https://raw.githubusercontent.com/michaellevine00/mit_intro_to_deep_learning/main/images/neural_network_sigmoid_activation.png)

We will use `torch.nn.Module` to define layers -- the building blocks of neural networks. Layers implement common neural networks operations. In PyTorch, when we implement a layer, we subclass `nn.Module` and define the parameters of the layer as attributes of our new class. We also define and override a function `forward`, which will define the forward pass computation that is performed at every step. All classes subclassing `nn.Module` should override the forward function.

Let's write a dense layer class to implement a perceptron defined above.



```python
### Defining a dense layer ###

# num_inputs: number of input nodes
# num_outputs: number of output nodes
# x: input to the layer

class OurDenseLayer(torch.nn.Module):
    def __init__(self, num_inputs, num_outputs):
        super(OurDenseLayer, self).__init__()
        # Define and initialize parameters: a weight matrix W and bias b
        # Note that the parameter initialize is random!
        self.W = torch.nn.Parameter(torch.randn(num_inputs, num_outputs))
        self.bias = torch.nn.Parameter(torch.randn(num_outputs))

    def forward(self, x):
        '''TODO: define the operation for z (hint: use torch.matmul).'''
        z = torch.matmul(x,self.W)+self.bias

        '''TODO: define the operation for out (hint: use torch.sigmoid).'''
        y = torch.sigmoid(z)
        return y
```

## How a dense layer is constructed
In the above code block, what is happening is the following: 

* **Defining `OurDenseLayer`**
  * We define a new class called **OurDenseLayer** by subclassing `torch.nn.Module` 

* **Initialization**
  * This layer is initialized with two arguments, `num_inputs`, and `num_outputs`

* **Weights and Biases**
  * Upon initialization, a matrix of weights and biases are created using the two arguments `num_inputs` and `num_outputs`
  * `self.W` is the weights matrix of the dense layer.  The `num_inputs` and `num_outputs` arguments inform the shape of the weights matrix, using the `torch.randn()` method to randomize the weights matrix.  The randomized matrix is then wrapped in the `torch.nn.Paramater()` method which ensures the weights matrix can learn and be updated, i.e. be a *learnable set of parameters*.  
  * The `self.W` weights matrix has the shape of `(num_inputs, num_outputs)` because this is a dense, fully connected layer.  For example, if there are two inputs and three outputs for our network, it is necessary for the weights matrix to be a 2 rows x 3 columns.  In other words, there are six weights.  Input 1 gets a unique weight for each of the three output neurons.  Input 2 gets a unique weight for each of the three output neurons. For example, \\(x_1*w_1\\) goes to neuron one.  

  \\(x_1*w_2\\) goes to neuron two, and so forth.

  * The `self.bias` matrix has the shape of `(num_outputs)` because each output layer will get a bias.  For example, if there are three outputs for our network, the bias matrix will be a 1-dimensional matrix of one row of three numbers.  After the three neurons compute and add the input values multiplied by an individual weight, a bias is added to that sum.  As with `self.W`, the `self.bias` matrix is a set of random numbers, and is wrapped in the `torch.nn.Paramater()` method so that the bias parameters can become learnable.  

  **Forward Pass**
  * The next part of the code defines how the neural network and passess the input(s) forward through the network to get outputs and specifically identifies the mathematical calculations that will occur.  
  * The `forward` pass of `OurDenseLayer` occurs after it is instantiated, and takes as an argument `x`.  `x` can be a sclar or a tensor.  The forward pass computes, for each output, a `z` value, which is a `torch.matmul` (matrix multiplication) operation. Specifically, this neural network multiples the input(s) by the weight matrix `self.W`, and adds to these the bias `self.bias`.  
  * The forward pass then takes this value and passes each output to the sigmoid function (a non-linearity), to get a result between 0 and 1.  

  **Return**
  * The function returns as an output a scalar or a tensor.
  
 
 




```python
### Defining a dense layer ###

# num_inputs: number of input nodes
# num_outputs: number of output nodes
# x: input to the layer

class OurDenseLayer(torch.nn.Module):
    def __init__(self, num_inputs, num_outputs):
        super(OurDenseLayer, self).__init__()
        # Define and initialize parameters: a weight matrix W and bias b
        # Note that the parameter initialize is random!
        self.W = torch.nn.Parameter(torch.randn(num_inputs, num_outputs))
        self.bias = torch.nn.Parameter(torch.randn(num_outputs))

    def forward(self, x):
        '''TODO: define the operation for z (hint: use torch.matmul).'''
        z = torch.matmul(x,self.W)+self.bias

        '''TODO: define the operation for out (hint: use torch.sigmoid).'''
        y = torch.sigmoid(z)
        return y
```
